```python
import matplotlib.pyplot as plt
from pathlib import Path
import pandas as pd
import os
import json
```

```python
reports_path = Path(os.path.dirname(os.getcwd())) / "reports"
metrics_path = reports_path / "insights" / "LightGBM_metrics.json"
figures_path = reports_path / "figures"
```

## Check model metrics

```python
with open(metrics_path, "r") as file:
    metrics = json.load(file)

metrics = pd.DataFrame(metrics, index=["Score"]).T
metrics
```

## Plot Three Model Graphs

```python
# Load the plots from the files
accuracy_over_samples_plot = plt.imread(figures_path / "LightGBM_Accuracy_Over_Samples.png")
confusion_matrix_plot = plt.imread(figures_path / "LightGBM_Confusion_Matrix.png")
historical_accuracy_plot = plt.imread(figures_path / "LightGBM_Historical_Accuracy.png")

# Display the plots
fig, axes = plt.subplots(3, 1, figsize=(15, 15))

# Plot 1: Accuracy over time
axes[0].imshow(accuracy_over_samples_plot)
axes[0].set_title('Accuracy Over Time')
axes[0].axis('off')

# Plot 2: Confusion matrix
axes[1].imshow(confusion_matrix_plot)
axes[1].set_title('Confusion Matrix')
axes[1].axis('off')

# Plot 3: Historical accuracy
axes[2].imshow(historical_accuracy_plot)
axes[2].set_title('Historical Accuracy')
axes[2].axis('off')

plt.tight_layout()
plt.show()
```
