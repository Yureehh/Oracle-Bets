```python
from pathlib import Path
import pandas as pd
import json
import os
```

```python
top_path = Path(os.path.dirname(os.getcwd()))
data_path = top_path / "data"
team_data_path = data_path / "processed" / "team_data.parquet"
```

```python
start_week_of_interest = "2023-10-31"
start_week_of_interest_dt = pd.to_datetime(start_week_of_interest)
end_week_of_interest_dt = start_week_of_interest_dt + pd.DateOffset(weeks=1)
```

```python
df = pd.read_parquet(team_data_path)

# Take all games where "date" is between start_week_of_interest and end_week_of_interest
df["date"] = pd.to_datetime(df["date"])
df = df[(df["date"] >= start_week_of_interest_dt) & (df["date"] < end_week_of_interest_dt)]

# Extract count of "league" played during that week
league_count = df["league"].value_counts().to_dict()
```
